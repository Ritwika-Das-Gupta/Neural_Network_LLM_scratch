```python
# build the vocabulary of characters and mappings to/from integers
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
print(itos)
```

```python

# build the dataset

block_size = 3 # context length: how many characters do we take to predict the next one?
X, Y = [], [] # X are the input to the neural net, Y are the labels
# iterating over all the words
for w in words:
  
  #print(w)
  # first we make a padded context of 0s
  context = [0] * block_size
  # then we iterate over the characters in the word
  # why are we adding a dot at the end?
# because we want to predict the end of the word too
  for ch in w + '.':
    ix = stoi[ch]
    # then we  append context of current inex
    X.append(context)
    # append the current charachter inex
    Y.append(ix)
    print(''.join(itos[i] for i in context), '--->', itos[ix])
    # then we update the context by removing the first character and adding the current character
    # this is like a sliding window of context
    context = context[1:] + [ix] # crop and append
  
X = torch.tensor(X)
Y = torch.tensor(Y)
```

```python
# creating a look up table for 2 dim
C = torch.randn((27, 2))
C
```

```python
# in previous lecture we encoded the X using one hot but that would be comlecated here
# so we can directly index frm lookup table which has the embedding of each X
```

```python
# to index we can directly give the tensor of X to C and that works in torch to give
# embeddig of each 32X3 intergers of X

emb = C[X]
emb.shape
```

```python
# here we actually dont need 3 dim we need 2 dim space so we use unbind to remove one dim and sqash it together
# 3, 2 squashed gives 6.. 1 is dimension to be squashed so torh .unbingd here gives 3 as 3 is to be squashed
# then we concat all to get final 2d tensor
# but this method is less efficient as it creates another tensor or another storage
inp=torch.cat(torch.unbind(emb, 1),1)
inp.shape
```

```python
# another most efficient function as explained in below  blog is view which will flatten any torch tesors
# exactly how we wish to view them
# here we give the dim how we wish to view and thats taken
emb.view(228146,6)
# now we can use this embedding and write our update statement
```

```python
# to understad how pytorch tensors are stored in computer memeory go though
# https://blog.ezyang.com/2019/05/pytorch-internals/
```

```python
# creating random weights no of  inputs = 3X2 3dim embedding and 2 such embeddings 
# as here last 2 dim are torch.Size([228146, 3, 2])
# no of neurons can be asumed anything here we take 100
# similarly 100 bias for each neuron
W1 = torch.randn((6, 100))
b1 = torch.randn(100)
```

```python
# logits
# -1 will help pytorch to derive the first dim whn nxt dim is already set 
# helps to remove the manual assigning
emb.view(-1, 6) @ W1 + b1
```

```python
# we use this logits layer frm first layer to make hidden layer with tanh as in diagram of MLP 

h = torch.tanh(emb.view(-1, 6) @ W1 + b1)
```

```python
h.shape
```

```python
#h is hidden layer of activation for all examples now
```

```python
# making the final layer
# input will be 100 coming from 00 neurons of hidden layer and output will be 27 alphabets embedding which will be our y_pred
# bias will be 27
W2 = torch.randn((100, 27))
b2 = torch.randn(27)
```

```python
# calclated from hidden layer using weights and bias of next layer
logits = h @ W2 + b2
```

```python
logits.shape
```

```python
#now we are done with 2 layers got logits so as per prev vedio we count and find prob
# and take -ve log likelihood
counts = logits.exp()
```

```python
prob = counts / counts.sum(1, keepdims=True)
```

```python

prob.shape
```

```python
# -ve log likelihood avg loss
loss = -prob[torch.arange(228146), Y].log().mean()
loss
```
